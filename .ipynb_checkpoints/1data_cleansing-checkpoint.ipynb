```python
import pandas as pd
import pandas
import sklearn.model_selection 
```

```python
#read the data 
df = pandas.read_csv("./raw_data/bank-full.csv", sep=";")
```

```python
display(df.head())
```

```python
summary = pd.DataFrame({
    "type": df.dtypes,
    "unique": df.nunique(dropna=False),
    "missing%": df.isna().mean() * 100,
    "unknown_%": df.eq("unknown").mean() * 100
})

print(summary.round(2).to_string())
```

```python
##drop colm duration to be fair model
df = df.drop(columns=['duration'])
df
```

```python
#create previously_contacted columns for handle -1 in pdays 
df["previously_contacted"] = (df["pdays"] != -1).astype(int)
df["pdays"] = df["pdays"].replace(-1, 0)
df
```

```python
# 4. ลบ unknown ใน job/education (ใส่ .copy() เพื่อป้องกัน SettingWithCopyWarning)
df = df[(df['job'] != 'unknown') & (df['education'] != 'unknown')].copy()
df
```

```python
# 5. Replace 'unknown' ใน contact และ poutcome
df['contact'] = df['contact'].replace('unknown', 'unrecorded')
df['poutcome'] = df['poutcome'].replace('unknown', 'nonexistent')
df['poutcome'] = df['poutcome'].replace('other', 'nonexistent')
df
```

```python
summary = pd.DataFrame({
    "type": df.dtypes,
    "unique": df.nunique(dropna=False),
    "missing%": df.isna().mean() * 100,
    "unknown_%": df.eq("unknown").mean() * 100
})

print(summary.round(2).to_string())
```

```python
# 6. แปลง Target 'y' เป็น 0 และ 1
df['y'] = df['y'].map({'no': 0, 'yes': 1})

# 7. ทำ One-Hot Encoding แปลงข้อความทั้งหมดเป็นตัวเลข
df_encoded = pd.get_dummies(df, drop_first=True)

# 8. แยก Features (X) และ Target (y)
X = df_encoded.drop(columns=['y'])
y = df_encoded['y']
```

```python
#แบ่งข้อมูล Train/Test
train_df, test_df = sklearn.model_selection.train_test_split(
    df, train_size=0.80, shuffle=True, random_state=0
)
```

```python
# save to csv 
train_df.to_csv("./cleaned_data/train_data.csv", index=False)
test_df.to_csv("./cleaned_data/test_data.csv", index=False)
```
