```python
import pandas as pd
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, accuracy_score, confusion_matrix

# 1. โหลดข้อมูล Train และ Test ที่ทำความสะอาดแล้ว
train_df = pd.read_csv("./cleaned_data/train_data.csv")
test_df = pd.read_csv("./cleaned_data/test_data.csv")

# 2. แยก Features (X) และ Target (y)
X_train = train_df.drop(columns=['y'])
y_train = train_df['y']

X_test = test_df.drop(columns=['y'])
y_test = test_df['y']

print(f"X_train shape: {X_train.shape}")
print(f"X_test shape: {X_test.shape}")

# 3. ทำ One-Hot Encoding
X_train_encoded = pd.get_dummies(X_train, drop_first=True)
X_test_encoded = pd.get_dummies(X_test, drop_first=True)

# 4. จัดคอลัมน์ของ Test ให้ตรงกับ Train (คอลัมน์ไหนขาดไปให้เติม 0)
X_train_encoded, X_test_encoded = X_train_encoded.align(X_test_encoded, join='left', axis=1, fill_value=0)

print(f"X_train_encoded shape: {X_train_encoded.shape}")
print(f"X_test_encoded shape: {X_test_encoded.shape}")

# 5. สร้างและเทรนโมเดล (ใช้ class_weight='balanced' เพื่อช่วยเรื่อง Imbalanced Data)
rf_model = RandomForestClassifier(n_estimators=100, random_state=0, class_weight='balanced')
rf_model.fit(X_train_encoded, y_train)

print("เทรนโมเดลเสร็จสมบูรณ์!")

# 6. ทำนายผลข้อมูล Test
y_pred = rf_model.predict(X_test_encoded)

# 7. แสดงผลลัพธ์
print("==========================================")
print("Accuracy Score:", accuracy_score(y_test, y_pred))
print("==========================================")

print("\nConfusion Matrix:\n", confusion_matrix(y_test, y_pred))
print("\nClassification Report:\n", classification_report(y_test, y_pred))

# 8. ดูความสำคัญของตัวแปร (Feature Importances)
importances = pd.DataFrame({
    'Feature': X_train_encoded.columns,
    'Importance': rf_model.feature_importances_
}).sort_values(by='Importance', ascending=False)

print("\nTop 5 Most Important Features:\n", importances.head(5))
```
