```python
import pandas as pd
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, accuracy_score, confusion_matrix
from sklearn.model_selection import GridSearchCV
from sklearn.preprocessing import StandardScaler

# 1. โหลดข้อมูล
train_df = pd.read_csv("../cleaned_data/train_data.csv")
test_df = pd.read_csv("../cleaned_data/test_data.csv")

# 2. Feature Engineering
def engineer_features(df):
    df = df.copy()
    df['is_senior'] = (df['age'] >= 60).astype(int)
    df['has_free_cash'] = ((df['balance'] > 0) & (df['housing'] == 'no') & (df['loan'] == 'no')).astype(int)
    df['total_contacts'] = df['campaign'] + df['previous']
    df['is_balance_negative'] = (df['balance'] < 0).astype(int)
    return df

train_fe = engineer_features(train_df)
test_fe = engineer_features(test_df)

X_train = train_fe.drop(columns=['y'])
y_train = train_fe['y']
X_test = test_fe.drop(columns=['y'])
y_test = test_fe['y']

# 3. One-Hot Encoding
X_train_encoded = pd.get_dummies(X_train, drop_first=True)
X_test_encoded = pd.get_dummies(X_test, drop_first=True)
X_train_encoded, X_test_encoded = X_train_encoded.align(X_test_encoded, join='left', axis=1, fill_value=0)

# 4. Feature Scaling (สำคัญมากสำหรับโมเดลตระกูล Linear)
scaler = StandardScaler()
X_train_scaled = pd.DataFrame(scaler.fit_transform(X_train_encoded), columns=X_train_encoded.columns)
X_test_scaled = pd.DataFrame(scaler.transform(X_test_encoded), columns=X_test_encoded.columns)

# 5. จูน Hyperparameter ด้วย GridSearchCV
param_grid = {
    'C': [0.01, 0.1, 1.0, 10.0], # ควบคุมความเข้มงวดของ Regularization
    'penalty': ['l2']
}

lr_base = LogisticRegression(random_state=0, class_weight='balanced', max_iter=1000)

grid_search = GridSearchCV(
    estimator=lr_base, 
    param_grid=param_grid, 
    cv=5, 
    scoring='f1', 
    n_jobs=-1
)

print("กำลังเทรน Logistic Regression...")
grid_search.fit(X_train_scaled, y_train)
best_lr = grid_search.best_estimator_

# 6. ประเมินผล
y_pred = best_lr.predict(X_test_scaled)

print("\n=== LOGISTIC REGRESSION (LINEAR) REPORT ===")
print("Accuracy:", accuracy_score(y_test, y_pred))
print("\nConfusion Matrix:\n", confusion_matrix(y_test, y_pred))
print("\nClassification Report:\n", classification_report(y_test, y_pred))

# 7. ดูปัจจัยที่ส่งผลบวกและลบมากที่สุด (Coefficients)
coefs = pd.DataFrame({
    'Feature': X_train_scaled.columns,
    'Weight': best_lr.coef_[0]
}).sort_values(by='Weight', ascending=False)

print("\nTop 5 ปัจจัยที่ทำให้ลูกค้าตอบ 'ตกลง' (Positive):\n", coefs.head(5))
print("\nTop 5 ปัจจัยที่ทำให้ลูกค้า 'ปฏิเสธ' (Negative):\n", coefs.tail(5))

import joblib
import os

os.makedirs("../models", exist_ok=True)
joblib.dump(best_lr, "../models/lr_model.pkl")
print("✅ บันทึก Logistic Regression เรียบร้อย!")
```
